```python
# Rebuild clean dataframe from original data
df_clean = df.copy()

# 1. Remove unnecessary index
df_clean = df_clean.drop(columns=["Unnamed: 0"])

# 2. Remove temporary EDA columns
eda_columns = [
    "Age_Group",
    "Income_Group",
    "DebtRatio_Group",
    "Utilization_Group",
    "Any_Delinquency",
    "90PlusLate"
]

df_clean = df_clean.drop(columns=eda_columns)

# 3. Remove invalid ages
df_clean = df_clean[
    (df_clean["age"] > 0) &
    (df_clean["age"] <= 100)
]

# 4. Remove exact duplicates
df_clean = df_clean.drop_duplicates()

# 5. Delinquency columns
delinquency_cols = [
    "NumberOfTime30-59DaysPastDueNotWorse",
    "NumberOfTime60-89DaysPastDueNotWorse",
    "NumberOfTimes90DaysLate"
]

# 6. Create delinquency anomaly flag BEFORE replacing 96/98
df_clean["Delinquency_Code_Flag"] = (
    df_clean[delinquency_cols]
    .isin([96, 98])
    .any(axis=1)
    .astype(int)
)

# 7. Replace 96/98 with missing values
for col in delinquency_cols:
    df_clean[col] = df_clean[col].replace([96, 98], np.nan)

# 8. Missingness indicators
df_clean["Income_Missing"] = (
    df_clean["MonthlyIncome"].isna()
).astype(int)

df_clean["Dependents_Missing"] = (
    df_clean["NumberOfDependents"].isna()
).astype(int)

# 9. IMPORTANT: create DebtRatio anomaly flag BEFORE replacing values
df_clean["DebtRatio_Anomaly"] = (
    df_clean["DebtRatio"] > 100
).astype(int)

# 10. Replace extreme DebtRatio values with NaN
df_clean.loc[
    df_clean["DebtRatio"] > 100,
    "DebtRatio"
] = np.nan

# 11. Utilization anomaly flag
df_clean["Utilization_Anomaly"] = (
    df_clean["RevolvingUtilizationOfUnsecuredLines"] > 10
).astype(int)

# Final checks
print("Shape:", df_clean.shape)
print("Duplicates:", df_clean.duplicated().sum())
print("DebtRatio anomalies:", df_clean["DebtRatio_Anomaly"].sum())
print("DebtRatio missing:", df_clean["DebtRatio"].isna().sum())
print("Utilization anomalies:", df_clean["Utilization_Anomaly"].sum())
```

```python
numeric_features = [
    "age",
    "Utilization_Log",
    "MonthlyIncome_Log",
    "DebtRatio_Log",
    "NumberOfOpenCreditLinesAndLoans",
    "NumberRealEstateLoansOrLines",
    "Total_Open_Loans",
    "NumberOfDependents",
    "NumberOfTime30-59DaysPastDueNotWorse",
    "NumberOfTime60-89DaysPastDueNotWorse",
    "NumberOfTimes90DaysLate",
    "Total_Delinquency_Count",
    "Any_Delinquency",
    "Has_90Plus_Delinquency",
    "Delinquency_Code_Flag",
    "Income_Missing",
    "Dependents_Missing",
    "DebtRatio_Anomaly",
    "Utilization_Anomaly",
    "High_Utilization_Flag"
]

categorical_features = ["Age_Group"]
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])
```

```python
logistic_model.fit(X_train, y_train)
```

```python
y_pred = logistic_model.predict(X_test)

y_prob = logistic_model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import (
    roc_auc_score,
    classification_report,
    confusion_matrix
)

print("ROC-AUC:", roc_auc_score(y_test, y_prob))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import numpy as np
import matplotlib.pyplot as plt

# ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# AUC
auc = roc_auc_score(y_test, y_prob)

# KS Statistic
ks = np.max(tpr - fpr)

# Gini
gini = 2 * auc - 1

print("ROC-AUC:", round(auc, 4))
print("KS Statistic:", round(ks, 4))
print("Gini Coefficient:", round(gini, 4))
```

```python
plt.figure(figsize=(8, 5))

plt.plot(fpr, tpr, label=f"Logistic Regression (AUC = {auc:.3f})")
plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Logistic Regression")
plt.legend()

plt.show()
```
